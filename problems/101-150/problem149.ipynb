```python
import numpy as np
import math
```

```python
grid_size = 100
```

```python
nums = [0] + ([None] * grid_size*grid_size)
def lag_fib_gen(k):
    if nums[k] is None:
        if k <= 55:
            nums[k] = (100003 - 200003*k + 300007*k*k*k) % 10**6 - 500000
        else:
            nums[k] = (lag_fib_gen(k-24) + lag_fib_gen(k-55) + 1000000) % 1000000 - 500000
    
    return nums[k]

for k in range(1, grid_size*grid_size + 1):
    lag_fib_gen(k)
```

```python
grid = np.zeros((grid_size+1, grid_size+1))
for i in range(1, grid_size+1):
    for j in range(1, grid_size+1):

        grid[i, j] = nums[grid_size*(i-1) + j]
```
