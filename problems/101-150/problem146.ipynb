Initial brute force was pretty good (got the answer in about 45 minutes). However, using the properties of primes given, we can deduce a few things:
1. $n$ must be even since $n^2 + 1$ is prime. From this note that $n$ must end in $0, 2, 4, 6,$ or $8$.
2. $n^2$ must end in a $0$, since $n^2 + 1$, $n^2 + 3$, $n^2 + 7$, and $n^2 + 9$ are all prime. Try the other ending digits and you will see that at least one of the those $4$ will end in $5$ (i.e., make it not prime, since $n^2 >= 100$).
3. Other remainder properties follow using math on residues for many small primes.

Essentially this reduces the search area to multiples of $10$, and they have to have particular remainders when divided by small primes.

```python
def miller_rabin(n, k = 10):
    if n <= 3:
        return n == 2 or n == 3

    if n % 2 == 0:
        return False

    r, s = 0, n - 1
    while s % 2 == 0:
        r += 1
        s //= 2
    for _ in range(k):
        a = random.randrange(2, n - 1)
        x = pow(a, s, n)
        if x == 1 or x == n - 1:
            continue
        for _ in range(r - 1):
            x = pow(x, 2, n)
            if x == n - 1:
                break
        else:
            return False
    return True
```

```python
sols = []
special_set = (1,3,7,9,13,27)
for n in range(10,150*10**6,10):
    if (n % 3 == 0) or not (3 <= n % 7 <= 4) or not (n % 13 in [1,3,4,9,10,12]) or (n % 17 in [2, 4, 5, 12 ,13, 15]) or (n % 19 in [4, 5, 7, 12, 14, 15]) or (n % 23 in [4, 19]): continue
    
    to_check = n*n

    works = True
    for i in range(1,28):
        if (i in special_set) != (miller_rabin(to_check + i)):
            works = False
            break
    
    if works:
        s = f'{n}: '
        for i in special_set:
            s += f'{to_check + i}, '
        
        sols.append(n)
        print(s)
```
